# Play the published ChessFly model

This notebook embeds the published ChessFly demo. The board is the pretrained FlyWire model: 138,639 neurons and 15,091,983 connections. You play White. The fly plays Black.

Playing a move is inference. It does not train the model, and it does not change the weights.

Run the next cell, then wait for the board. The first visit downloads about 148 MB into your browser. Later visits can use the browser cache.

This is a different brain map from the MaleCNS v1.0 excerpt in the fruit-fly notebook.


# 1. Open the board

The cell below shows the published site inside this notebook. The kernel does not download the brain and does not choose a move. Your browser does both.

The files come from the author's public Space:

`https://mlabonne-chessfly.static.hf.space/`

That host has the page, the inference worker, the packed connectome, and `flynet.safetensors`. Leave this tab open until the status line names the neuron and connection counts. Then move a White piece.


```python
from IPython.display import IFrame, display

# The huggingface.co/spaces page sends X-Frame-Options: DENY.
# The static Space host is the page that can sit inside a notebook.
display(IFrame(
    src="https://mlabonne-chessfly.static.hf.space/",
    width=980,
    height=820,
))
```

# 2. What this uses

**Computation.** The forward pass runs in your browser. The published worker tries WebGPU, and otherwise runs the same math in JavaScript. A depth-3 search in that page calls the same forward pass more than once. None of that runs in the DataHub kernel.

**Model files.** The browser loads about 148 MB from the Space: `data/connectome.bin.gz` (47,051,343 bytes), `data/neurons.bin.gz` (1,442,541 bytes), and `data/flynet.safetensors` (99,572,154 bytes). This repository does not contain those files.

**Internet.** Your browser needs access to `mlabonne-chessfly.static.hf.space` and the Hugging Face file host it redirects to. The kernel does not need that network. After a successful download, the app says the files are cached in the browser.

**Training.** The published worker accepts `init` and `evaluate` only. There is no training control on this board.


# 3. Permissions, kept separate

**App.** The Space card says GPL-3.0 because the page includes chessground. This notebook does not copy that program. It links to the copy the author already hosts. GPL-3.0's conditions apply if someone later conveys that program, or a modified version of it, for example by committing the Space JavaScript to this repository: keep the license notice, and make the corresponding source available. Those are conditions on distribution. They are not a reason to avoid the hosted demo.

**Weights.** The model card says `license: other`. The model repository has no separate license file and no other grant text. Loading `flynet.safetensors` from the author's public Space is how that demo is published. Copying the weight file into this course repository is not covered by a stated license, so this prototype does not do that.

**Connectome.** FlyWire's public-release guidelines say the public release data is available under CC BY-NC 4.0. The model card says those non-commercial terms apply to anything derived from the graph. The packed `connectome.bin.gz` in the Space is that kind of file. This notebook does not host it. The Space metadata records a hash for an upstream `LICENSE` file and does not include the file text, so any terms in that author-release license beyond the FlyWire public-release terms are still unread. CC BY-NC 4.0, where it applies, asks for attribution, a link to the license, and non-commercial use. Attribution is in Sources.

**Labels.** The positions used to train the published weights came from Lichess/chess-position-evaluations, which is CC0. This notebook does not ship those positions.
